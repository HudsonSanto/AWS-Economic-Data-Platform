```python
#Importação Bibliotecas

import boto3
import requests
import pandas as pd 

```

```python
#Importação de DataFrame

url = "https://api.bcb.gov.br/dados/serie/bcdata.sgs.433/dados"

params = {
    "formato": "json"
}

response = requests.get(url, params=params)

dados = response.json()

df = pd.DataFrame(dados)

df.sort_values(by='data', ascending=False).head(7)
```

```python
#Transformação Dados

df['data'] = pd.to_datetime(
    df['data'],
    format='%d/%m/%Y'
)

df['valor'] = df['valor'].astype(float)

df.info()
```

```python
#Transformar para parquet

df.to_parquet(
    "ipca.parquet",
    index=False
)
```

```python
#Integração com client AWS
bucket_name = "econimiccentralbank"

s3 = boto3.client("s3")

s3.upload_file(
    "ipca.parquet",
    bucket_name,
    "raw/ipca/ipca.parquet"
)

print(f'arquivo salvo no bucket {bucket_name}')
```
